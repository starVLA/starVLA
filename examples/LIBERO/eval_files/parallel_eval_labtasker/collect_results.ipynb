```python
# Filter by benchmark to avoid mixing with RoboTwin tasks
# Add more conditions as needed:
# e.g. extra_filter = 'metadata.benchmark == "LIBERO" and metadata.ckpt_stem == "steps_30000_pytorch_model"'
# e.g. extra_filter = 'metadata.benchmark == "LIBERO" and args.task_suite == "libero_spatial"'
extra_filter = 'metadata.benchmark == "LIBERO"'

tasks = ls_tasks(
    status="success",
    extra_filter=extra_filter if extra_filter else None,
    limit=1000,
).content
print(f"{len(tasks)} successful tasks")
```

```python
import pandas as pd

rows = []
for t in tasks:
    rows.append({
        "ckpt_stem": t.metadata.get("ckpt_stem", ""),
        "task_suite": t.summary["task_suite"],
        "total_episodes": t.summary["total_episodes"],
        "total_successes": t.summary["total_successes"],
        "success_rate": t.summary["success_rate"],
    })

df = pd.DataFrame(rows)
df = df.set_index(["ckpt_stem", "task_suite"]).sort_index()
df
```
